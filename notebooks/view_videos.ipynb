# 動画の表示

フォルダの中の mp4 をノートブック上で再生する（ダウンロード不要）。標準のカーネル（Python 3）でも動く。
`FOLDER` を見たいフォルダ（リポジトリからの相対パス）に変えて実行する。

| 見たいもの | FOLDER の例 |
|---|---|
| 生成したデータ（`scripts/export_dataset_videos.sh` で書き出したもの） | `outputs/dataset_videos/pot_multi_v1` |
| 評価の動画（`scripts/evaluate.sh` の結果） | `outputs/eval/act_pot_cube_v1/videos` |


```python
FOLDER = "outputs/dataset_videos/pot_multi_v1"

from pathlib import Path

from IPython.display import Markdown, Video, display

repo = Path.cwd().resolve()
if repo.name == "notebooks":
    repo = repo.parent
videos = sorted((repo / FOLDER).glob("*.mp4"))
print(f"{repo / FOLDER}: {len(videos)} 本")
for path in videos:
    display(Markdown(f"**{path.name}**"))
    display(Video(str(path), embed=True, width=560))
```
