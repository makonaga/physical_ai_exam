# 評価動画の表示

`outputs/eval/<NAME>/videos/` の動画をノートブック上で再生する（ダウンロード不要）。
標準のカーネル（Python 3）でも動く。`NAME` を見たい評価の名前に変えて実行する。
左が俯瞰カメラ、右が手首カメラ。ファイル名の末尾が結果（success / not_grasped / missed_pot）。

```python
NAME = "act_pot_cube_v1"

from pathlib import Path

from IPython.display import Markdown, Video, display

repo = Path.cwd().resolve()
if repo.name == "notebooks":
    repo = repo.parent
videos = sorted((repo / "outputs" / "eval" / NAME / "videos").glob("*.mp4"))
print(f"{len(videos)} 本")
for path in videos:
    display(Markdown(f"**{path.name}**"))
    display(Video(str(path), embed=True, width=448))
```
